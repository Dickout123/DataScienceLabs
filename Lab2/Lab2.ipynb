```python
import numpy as np

# Створення набору даних (розмір за власним рішенням, наприклад, 10 елементів)
array_size = 10

# 1. Створення лінійних значень за допомогою np.arange
# Створюємо вектор-скаляр від 0 до 10 (не включно)
linear_array = np.arange(array_size)
print("1. Лінійний вектор (arange):")
print(linear_array)

# 2. Створення випадкових чисел за допомогою np.random
# Створюємо вектор випадкових чисел у діапазоні [0, 1)
random_array = np.random.rand(array_size)
print("\n2. Випадковий вектор (random):")
print(random_array)

# 3. Створення матриці (вектор 3x4)
matrix = np.random.randint(1, 10, size=(3, 4))
print("\n3. Матриця (3x4):")
print(matrix)
```

```python
# READ (Читання/Отримання)
print("READ: Значення в матриці на позиції [1, 2]:", matrix[1, 2])
print("READ: Перший рядок матриці:", matrix[0, :])

# UPDATE (Оновлення)
# Змінимо значення у випадковому масиві (5-й елемент)
random_array[4] = 99.9
print("\nUPDATE: Випадковий масив після оновлення 5-го елемента:")
print(random_array)

# DELETE (Видалення)
# Видаляємо 2-й стовпець (індекс 1) з матриці
# Примітка: np.delete не змінює вихідний масив, а повертає новий
matrix_deleted_col = np.delete(matrix, 1, axis=1)
print("\nDELETE: Матриця після видалення 2-го стовпця:")
print(matrix_deleted_col)
```

```python
import pandas as pd

# 1. Створення DataFrame зі словника (для читабельності)
data_dict = {
    'Ім\'я': ['Олег', 'Наталя', 'Ігор', 'Марія'],
    'Вік': [25, 31, 22, 45],
    'Місто': ['Київ', 'Львів', 'Одеса', 'Київ']
}
df_from_dict = pd.DataFrame(data_dict)
print("1. DataFrame, створений зі словника:")
print(df_from_dict)

# 2. Створення DataFrame з масиву NumPy
# Створюємо NumPy масив 5x3
np_data = np.random.randint(1, 100, size=(5, 3))
df_from_numpy = pd.DataFrame(np_data, columns=['Коло A', 'Коло B', 'Коло C'])
print("\n2. DataFrame, створений з масиву NumPy:")
print(df_from_numpy)
```

```python
# 1. head() - Відображення перших рядків
print("1. Перші 3 рядки DataFrame зі словника (head(3)):")
print(df_from_dict.head(3))

# 2. describe() - Опис статистики (для числових колонок)
print("\n2. Опис статистики DataFrame з NumPy (describe()):")
print(df_from_numpy.describe())

# 3. iloc (Індексація за позицією)
# Отримати значення в 2-му рядку та 3-му стовпці (індекси 1, 2)
print("\n3. iloc: Елемент [1, 2] (DataFrame зі словника):")
print(df_from_dict.iloc[1, 2])

# 4. loc (Індексація за міткою/назвою)
# Отримати всі дані для рядка з індексом 2 (за назвою індексу)
print("\n4. loc: Рядок з індексом 2 (DataFrame зі словника):")
print(df_from_dict.loc[2])
```

```python
import matplotlib.pyplot as plt

# 1. Генерація даних для лінійної функції Y = 2X + 5
X = np.arange(0, 10, 0.5)  # Точки X
Y_true = 2 * X + 5         # Ідеальні значення Y

# 2. Додавання помилок вимірювань (випадковий шум)
# np.random.normal(0, 2, size=len(X)) генерує шум з середнім 0 та стандартним відхиленням 2
noise = np.random.normal(0, 2, size=len(X))
Y_data = Y_true + noise    # Отримані дані з шумом
```

```python
# Побудова графіка
plt.figure(figsize=(10, 6))

# Графік ідеальної функції
plt.plot(X, Y_true, label='Ідеальна функція ($Y=2X+5$)', color='red', linestyle='--')

# Графік згенерованих даних з помилками
plt.scatter(X, Y_data, label='Згенеровані дані (з шумом)', color='blue', marker='o')

plt.title('Лінійна функція та згенеровані дані')
plt.xlabel('Ось X')
plt.ylabel('Ось Y')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Обчислення метрик похибок
# Похибка — це різниця між істинним значенням та виміряним/прогнозованим
errors = Y_data - Y_true

# MAE (Середня абсолютна похибка)
# Середнє значення абсолютних помилок
mae = np.mean(np.abs(errors))

# MSE (Середня квадратична похибка)
# Середнє значення квадратів помилок
mse = np.mean(errors**2)

print(f"\nОбчислені метрики похибок:")
print(f"MAE (Mean Absolute Error): {mae:.4f}")
print(f"MSE (Mean Squared Error): {mse:.4f}")
```

```python
# Створення DataFrame для результатів
results_df = pd.DataFrame({
    'X': X,
    'Y_true': Y_true,         # Y - Ідеальні значення
    'Y_data': Y_data,         # Y_hat - Згенеровані дані з шумом (аналог прогнозу)
    'MAE': [mae] * len(X),    # Додаємо MAE до всіх рядків
    'MSE': [mse] * len(X)     # Додаємо MSE до всіх рядків
})

# Запис у CSV файл
csv_filename = 'results_lab2.csv'
results_df.to_csv(csv_filename, index=False)

print(f"\nДані успішно записано у файл: '{csv_filename}'")
```
